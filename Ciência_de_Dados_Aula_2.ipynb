```python
import pandas as pd
```

```python
#Criar um dataframe a partir de um dicionário
data = {'Nome': ['Alice', 'Bob', 'Carol'],
        'Idade': [25, 30, 22]}
df = pd.DataFrame(data)
print(df)
```

```python
serie_idade = df['Idade']
print(serie_idade)
```

```python
serie_nomes = df['Nome']
print(serie_nomes)
```

```python
serie_nomes.describe()
```

```python
jovens = df[df['Idade'] < 30]
print(jovens)
```

```python
#Adicionando uma nova coluna ao DataFrame:
df['Cidade'] = ['Barra Mansa', 'Volta Redonda', 'Barra do Piraí']
```

```python
print(df)
```

```python
#Removendo uma coluna do DataFrame
df.drop('cidade', axis=1, inplace=True)
```

```python
print(df)
```

```python
#Aplicar uma função a uma coluna
df['Idade_dobrada'] = df['Idade'].apply(lambda x: x * 2)
```

```python
print(df)
```

```python
#Remover linhas com valores ausentes
df_limpo = df.dropna()
```

```python
print(df_limpo)
```

```python
#Preencher valores ausentes
df_preenchido = df.fillna(0)
```

```python
print(df_preenchido)
```

```python
#Calcular a média da idade por cidade
media_idade_por_cidade = df.groupby('Cidade')['Idade'].mean()
```

```python
print(media_idade_por_cidade)
```

```python
#Criar outro DataFrame
data2 = {'Cidade': ['Nova York', 'Los Angeles'],
         'População': [8500000, 4000000]}
df2 = pd.DataFrame(data2)
```

```python
print(df2)
```

```python
#Combinar DataFrames usando uma chave
df_combinado = pd.merge(df, df2, on='Cidade', how = 'outer')
```

```python
print(df_combinado)
```
