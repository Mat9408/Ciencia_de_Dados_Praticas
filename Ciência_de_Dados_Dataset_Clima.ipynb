Também conseguimos fazer filtros na base

```python
#Verificando temperaturas maiores que 80
df[df.Temperatura > 80]
```

```python
#Verificando temperaturas menores que 70
df[df.Temperatura <70]
```

Usando filtros de E e OU

```python
#Verificando se há temperatura E umidade maiores que 80
df[(df.Temperatura > 80) & (df.Umidade > 80)].head()
```

Uma forma muito útil de fazer seleção de dados é usando o .loc() ou o .iloc()

```python
#O .loc() permite fazer a busca da mesma forma acima
df.loc[(df.Temperatura > 1) & (df.Umidade > 1)].head()
```

```python
#Ele também permite usar argumento lógicos
df.loc[(df.Temperatura > 1) | (df.Umidade > 1)].head()
```

```python
#Ele permite filtrar colunas de forma muito prática
df.loc[(df.Temperatura > 1) | (df.Umidade > 1), ['Aparência','Jogar']].head()
```

Já o .iloc() vai usar o índice para filtrar

```python
#Buscando as linhas de 3(inclusive) a 4(exclusive)
df.iloc[3:4]
```

```python
#Também posso usar para buscar apenas colunas específicas
df.iloc[3:7, 1:3]
```
