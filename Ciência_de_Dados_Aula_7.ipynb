```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Duas listas ou arrays
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 11]
plt.plot(x,y)
plt.title("Exemplo - Gráfico de Linhas")
plt.xlabel("Meses")
plt.ylabel("Valores")
plt.show()
```

```python
# Dicionário
dados = {
 'Janeiro': 10,
 'Fevereiro': 20,
 'Março': 15,
 'Abril': 25
}
plt.plot(dados.keys(), dados.values(), marker="o")
plt.title("Exemplo - Gráfico de Linhas com dicionário")
plt.xlabel("Meses")
plt.ylabel("Valores")
plt.show()
```

```python
#Gráfico de barras
categorias = ['A', 'B', 'C']
valores = [4, 7, 1]
plt.bar(categorias, valores)
plt.title("Gráfico de Barras")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()
```

```python
#Gráfico de Barras usando Dicionário
dados = {
 'Categoria A': 4,
 'Categoria B': 7,
 'Categoria C': 1
}
plt.bar(dados.keys(), dados.values())
plt.title("Gráfico de Barras a partir de Dicionário")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()
```

```python
#Gráfico de dispersão(scatterplot)
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
plt.scatter(x, y)
plt.title("Gráfico de Dispersão")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
dados_x = {
 'A': 1,
 'B': 2,
 'C': 3
}
dados_y = {
 'A': 2,
 'B': 3,
 'C': 5
}
plt.scatter(dados_x.values(), dados_y.values())
plt.title("Gráfico de Dispersão a partir de Dicionários")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.xticks(list(dados_x.values()), list(dados_x.keys()))
plt.show()
```

```python
#Gráfico de pizza
partes = [20, 30, 50]
rotulos = ['A', 'B', 'C']
plt.pie(partes, labels=rotulos, autopct='%1.1f%%')
plt.title("Gráfico de Pizza")
plt.show()
```

```python
#Gráfico de pizza em dicionário
dados = {
 'Parte A': 20,
 'Parte B': 30,
 'Parte C': 50
}
plt.pie(dados.values(), labels=dados.keys(), autopct='%1.1f%%')
plt.title("Gráfico de Pizza a partir de Dicionário")
plt.show()
```
